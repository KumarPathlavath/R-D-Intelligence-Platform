# Signal-to-Opportunity: Interactive R&D Intelligence Exploration
This notebook demonstrates exploratory data analysis on external signals, organizational capability embeddings, cosine similarity gap analysis, and time-windowed signal clusters.

```python
import os
import sys
import pandas as pd
import numpy as np
from src.storage.database import DatabaseManager
from src.capability.capability_store import CapabilityStore
from src.gap_analysis.similarity import GapAnalyzer
from src.gap_analysis.clustering import TemporalSignalClusterer
from src.evaluation.evaluate import EvaluationEngine

db = DatabaseManager()
print("Database Summary:", db.get_summary_stats())
```

## 1. Inspect Internal Organizational Capabilities

```python
caps = db.get_capabilities()
df_caps = pd.DataFrame([
    {
        "ID": c.capability_id,
        "Name": c.name,
        "Domain": ", ".join(c.domain),
        "TRL": c.trl,
        "Active Projects": len(c.active_projects)
    }
    for c in caps
])
df_caps
```

## 2. Inspect Surfaced Evidence-Backed Opportunities

```python
opps = db.get_opportunities()
df_opps = pd.DataFrame([
    {
        "ID": o.opportunity_id,
        "Title": o.title,
        "Domain": o.domain,
        "Confidence": f"{o.confidence_score*100:.1f}%",
        "Gap Type": o.gap_type.value,
        "Sources": len(o.supporting_sources),
        "Status": o.triage_status.value
    }
    for o in opps
])
df_opps
```
